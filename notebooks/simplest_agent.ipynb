```python
from agents import Agent, Runner, trace
from openai.types.responses import ResponseTextDeltaEvent
```

Create a Placement Trainer Agent to help students prepare for job placements.

```python
placement_trainer_agent = Agent(
    name="Placement Trainer",
    instructions="""
    You are a helpful guide for students preparing for job placements.
    Help with interview practice, resumes, aptitude tests, communication, and preparation plans.
    Give concise, practical answers with clear next steps.
    Tailor your advice to the student's target role, experience, and timeline when known.
    If important context is missing, ask one brief clarifying question.
    """,
)

```

Let's ask the agent for a practical placement-preparation plan:

```python
with trace("Placement Trainer Agent"):
    result = await Runner.run(
        placement_trainer_agent,
        "I have two weeks before campus placements. How should I prepare for software engineering interviews?",
    )

print(result)
```

Streaming a concise placement-interview tip to the screen, token by token

```python
response_stream = Runner.run_streamed(
    placement_trainer_agent,
    "Give me one concise tip for answering 'Tell me about yourself' in a placement interview.",
)

async for event in response_stream.stream_events():
    if event.type == "raw_response_event" and isinstance(
        event.data, ResponseTextDeltaEvent
    ):
        print(event.data.delta, end="", flush=True)
        

```

```python
student_manager_agent = Agent(
    name="Student Manager Agent",
    instructions="""
    You are a helpful manager who guides for students. 
    """,
)
```

```python
with trace("Student Manager Agent"):
    result = await Runner.run(
        student_manager_agent,
        "I have two weeks before campus placements. How should I prepare for software engineering interviews?",
    )

print(result)
```
